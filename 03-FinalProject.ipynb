## Research Question

Which combination of player performance statistics can be optimally integrated into a single aggregate metric that most accurately predicts the winners of the NHL regular-season awards (Hart Memorial Trophy, Vezina Trophy, and Norris Trophy)?

Specifically, this research seeks to determine how traditional and advanced statistical measures, such as goals, assists, plus/minus, minutes played, and games played, among others, can be weighted and combined to form a composite indicator of player value that best aligns with historical award outcomes.

## Skaters Analysis

### Plus/Minus Analysis

Since we think plus/minus will be the most important indicator for the Hart trophy winners, we are going to visualize it first, highlighting the player seasons where the trophy was won with red points.

```python
import pandas as pd
import matplotlib.pyplot as plt
import unicodedata

# Function to handle special characters (e.g., Pastrňák -> Pastrnak)
def normalize_name(name):
    if not isinstance(name, str): return name
    return "".join(c for c in unicodedata.normalize('NFD', name) if unicodedata.category(c) != 'Mn')

# 1. Load both datasets
skaters = pd.read_csv('skaters_2008_to_2024.csv')
harts = hart_data

# 2. Calculate Plus/Minus (Logic: 5v5 + SHG - 5v5 - SHGA)
skaters['plus_val'] = 0
skaters.loc[skaters['situation'].isin(['5on5', '4on5', 'other']), 'plus_val'] = skaters['OnIce_F_goals']
skaters['minus_val'] = 0
skaters.loc[skaters['situation'].isin(['5on5', '5on4', 'other']), 'minus_val'] = skaters['OnIce_A_goals']

pm_data = (
    skaters.groupby(['playerId', 'name', 'season'])
           .agg({'plus_val': 'sum', 'minus_val': 'sum'})
           .reset_index()
)
pm_data['plus_minus'] = pm_data['plus_val'] - pm_data['minus_val']

# 3. Extract full-season totals from the 'all' situation
totals = skaters[skaters['situation'] == 'all'][[
    'playerId', 'name', 'season',
    'I_F_points', 'I_F_goals', 'I_F_xGoals',
    'I_F_shotAttempts', 'icetime', 'games_played'
]].copy()

totals.rename(columns={
    'I_F_points': 'total_points',
    'I_F_shotAttempts': 'I_F_shots',
    'icetime': 'I_F_minutes'
}, inplace=True)

# 4. Merge Plus/Minus with full-season totals
final_df = pd.merge(
    pm_data[['playerId', 'name', 'season', 'plus_minus']],
    totals,
    on=['playerId', 'name', 'season'],
    how='inner'
)

# 5. Filter and Merge Hart Winners
harts_winners = harts[harts['Place'] == 1].copy()
harts_winners['norm_player'] = harts_winners['Player'].apply(normalize_name)
final_df['norm_name'] = final_df['name'].apply(normalize_name)

hart_stats = pd.merge(
    harts_winners,
    final_df,
    left_on=['norm_player', 'Year'],
    right_on=['norm_name', 'season']
)

# 6. Create the Visualization
plt.figure(figsize=(16, 10))

plt.scatter(final_df['total_points'], final_df['plus_minus'], 
            c='lightgrey', alpha=0.3, s=15, label='All Other Players')

plt.scatter(hart_stats['total_points'], hart_stats['plus_minus'], 
            c='red', s=100, edgecolors='black', linewidth=1.5, zorder=5, label='Hart Trophy Winners')

for i, row in hart_stats.iterrows():
    plt.annotate(f"{row['name']} ({row['season']})", 
                 (row['total_points'], row['plus_minus']),
                 textcoords="offset points", xytext=(5,5), 
                 fontsize=9, fontweight='bold', color='darkred')

plt.axhline(0, color='black', linewidth=1, alpha=0.3)
plt.xlabel('Total Points in Season', fontsize=12)
plt.ylabel('Plus/Minus (+/-)', fontsize=12)
plt.title('The Elite Circle: Hart Trophy Winners vs. The Field (2008-2024)', fontsize=16, fontweight='bold')
plt.grid(True, linestyle=':', alpha=0.5)
plt.legend(loc='lower right')

plt.show()
```

The winners tend to have a positive plus/minus, and do stand out from the crowd, yet there seem to be some notable outliers with an incredibly high plus/minus and points scored who did not win the award.

### Skaters Breakdown, Efficiency

The main objective of hockey is scoring goals, but there are a few ways to measure the efficiency with which players score. We will show how Hart winners stack up vs the rest of the field. 

#### Goals per 60 vs Expected Goals per 60

This is a metric that normalizes efficiency by time on the ice.

```python
import matplotlib.pyplot as plt

# Compute per-60 rates
final_df['goals_per60'] = final_df['I_F_goals'] / final_df['I_F_minutes'] * 60
final_df['xgoals_per60'] = final_df['I_F_xGoals'] / final_df['I_F_minutes'] * 60

hart_stats['goals_per60'] = hart_stats['I_F_goals'] / hart_stats['I_F_minutes'] * 60
hart_stats['xgoals_per60'] = hart_stats['I_F_xGoals'] / hart_stats['I_F_minutes'] * 60

plt.figure(figsize=(16,10))

# Background
plt.scatter(final_df['xgoals_per60'], final_df['goals_per60'],
            c='lightgrey', alpha=0.3, s=15)

# Hart winners
plt.scatter(hart_stats['xgoals_per60'], hart_stats['goals_per60'],
            c='red', s=100, edgecolors='black', linewidth=1.5, zorder=5)

# Labels
for _, row in hart_stats.iterrows():
    plt.annotate(f"{row['name']} ({row['season']})",
                 (row['xgoals_per60'], row['goals_per60']),
                 textcoords="offset points", xytext=(5,5),
                 fontsize=9, fontweight='bold', color='darkred')

plt.xlabel("Expected Goals per 60")
plt.ylabel("Goals per 60")
plt.title("Shot Quality vs. Shot Conversion: Hart Winners vs. The Field")
plt.grid(True, linestyle=':', alpha=0.5)
plt.show()
```

```python
skaters['is_goalie'] = 0
goalies_data['is_goalie'] = 1

# Use the actual column names you have
common_cols = [
    'playerId', 'season', 'name', 'team',
    'position', 'games_played', 'icetime', 'is_goalie'
]

skater_only_cols = [c for c in skaters.columns if c not in common_cols]
goalie_only_cols = [c for c in goalies_data.columns if c not in common_cols]

# Add missing goalie columns to skaters
missing_for_skaters = [c for c in goalie_only_cols if c not in skaters.columns]
if missing_for_skaters:
    skaters = pd.concat(
        [skaters, pd.DataFrame({c: np.nan for c in missing_for_skaters}, index=skaters.index)],
        axis=1
    )

# Add missing skater columns to goalies
missing_for_goalies = [c for c in skater_only_cols if c not in goalies_data.columns]
if missing_for_goalies:
    goalies_data = pd.concat(
        [goalies_data, pd.DataFrame({c: np.nan for c in missing_for_goalies}, index=goalies_data.index)],
        axis=1
    )

# Combine
all_cols = common_cols + skater_only_cols + goalie_only_cols

players_season = pd.concat(
    [skaters[all_cols], goalies_data[all_cols]],
    ignore_index=True
)

hart_data['season'] = hart_data['Year']
hart_data['hart_winner'] = (hart_data['Place'] == 1).astype(int)
hart_merge = hart_data[['season', 'Player', 'Tm', 'hart_winner', 'Vote%']].copy()
hart_merge = hart_merge.rename(columns={'Player': 'name', 'Tm': 'team'})

model_df = pd.merge(
    hart_merge,
    players_season,
    on=['season', 'name', 'team'],
    how='inner'
)
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import LogisticRegression
from sklearn.pipeline import Pipeline

target_col = 'hart_winner'

feature_cols = model_df.select_dtypes(include=['number']).columns.tolist()
feature_cols = [c for c in feature_cols if c not in ['hart_winner', 'Vote%']]

X = model_df[feature_cols].fillna(0)
y = model_df[target_col].astype(int)

X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=42, stratify=y
)

logit_clf = Pipeline([
    ('scaler', StandardScaler()),
    ('clf', LogisticRegression(max_iter=500, class_weight='balanced')),
])

logit_clf.fit(X_train, y_train)

from sklearn.metrics import roc_auc_score, accuracy_score

y_pred_proba = logit_clf.predict_proba(X_test)[:, 1]
y_pred = logit_clf.predict(X_test)

print("Accuracy:", accuracy_score(y_test, y_pred))
print("ROC AUC:", roc_auc_score(y_test, y_pred_proba))
```

```python
current_season = 2024

current_df = players_season[players_season['season'] == current_season].copy()

X_current = current_df[feature_cols].fillna(0)

current_df['hart_prob'] = logit_clf.predict_proba(X_current)[:, 1]

predictions = (
    current_df[['name', 'team', 'position', 'hart_prob']]
    .sort_values('hart_prob', ascending=False)
)

predictions.head(15)


top15 = predictions.head(15)

plt.figure(figsize=(10,6))
sns.barplot(data=top15, x='hart_prob', y='name', hue='team')
plt.title(f"Top 15 Predicted Hart Candidates – {current_season}")
plt.xlabel("Predicted Hart Win Probability")
plt.ylabel("Player")
plt.legend(title="Team")
plt.show()
```
